```python
# Generating results tables for various datasets

import sys
from pathlib import Path

# add parent folder to path
sys.path.append(str(Path("..").resolve()))  # goes one level up

import importlib
import proximitylib.proximity as prox
importlib.reload(prox)
from tabulate import tabulate

# --------------------------------------------------------------
# Function to show results
# --------------------------------------------------------------

def show_results(path, name, params, num_null=1000, alpha=0.1,  models=["expon", "uniform"]):
    """
    Analyzes a dataset peaks table and prints the summarized results.
    Parameters:
        path (str): Path to the dataset peaks table.
        name (str): Name of the dataset for display purposes.
        params (dict): Parameter dictionary for analysis.
        num_null (int): Number of null models to generate.
        alpha (float): Significance level for statistical tests.
        models (list): List of models to consider.
    Returns:
        df (DataFrame): Detailed results table.
        df_summary (DataFrame): Summarized results table.
        distance_corr_dict (dict): Dictionary of distance correlations.
    """
    df, distance_corr_dict = prox.analyze_dataset_peaks_table(path, params, num_null=num_null, alpha=alpha, models=models)
    #prox.print_peak_table(df, title=name)

    df_summary = prox.summarize_results_table(df, distance_corr_dict=distance_corr_dict, alpha=alpha)
    print(f"\n{name} (Summary)\n" + "="*len(f"{name} (Summary)"))
    print(tabulate(df_summary, headers="keys", tablefmt="fancy_grid", showindex=False))

    return df, df_summary, distance_corr_dict

# === Choose a Parameter-Set ===
param_choice = 3         # options 1, 2 or 3 
params = prox.param_sets[param_choice]
print(f"Applied Parameter-Dict: default_params_{param_choice}")

# Application:
accelerando = show_results("../data/ergebnis-tabellen/accelerando/accelerando.csv", "Accelerando", params=params)
drift = show_results("../data/ergebnis-tabellen/isochron drift/isochron_drift.csv", "Drift", params=params)
jitter_middle = show_results("../data/ergebnis-tabellen/jitter middle", "Jitter middle", params=params)
tempo_shift = show_results("../data/ergebnis-tabellen/tempo shift", "Tempo Shift", params=params)
heterochrony_1_2 = show_results("../data/ergebnis-tabellen/heterochrony-1-2", "Heterochrony 1-2", params=params)
heterochrony_2_3 = show_results("../data/ergebnis-tabellen/heterochrony-2-3", "Heterochrony 2-3", params=params)
isochrony_small_jitter = show_results("../data/ergebnis-tabellen/isochrony_small-jitter", "Isochrony Small Jitter", params=params)
isochrony_big_jitter = show_results("../data/ergebnis-tabellen/isochrony_big-jitter", "Isochrony Big Jitter", params=params)
random_exponential = show_results("../data/ergebnis-tabellen/random exponential", "Random Exponential", params=params)
random_uniform = show_results("../data/ergebnis-tabellen/random uniform", "Random Uniform", params=params)

C_persp = show_results("../data/Seba's short-tailed bat (Carollia perspicillata)", "Carollia perspicillata", params=params)
kwa = show_results("../data/Kwa-Sounds (Scorpaena spp.)", "Scorpaena spp.", params=params)
Sperm_whale = show_results("../data/Sperm whale (Physeter macrocephalus)", "Physeter macrocephalus", params=params)
```
