```python
# This script generates results tables for various datasets using the proximitylib library.
# It analyzes dataset peaks tables, summarizes the results, and prints them in a formatted table. 
# The script allows for different parameter sets to be applied to the analysis.

# Generating results tables for various datasets
import sys
from pathlib import Path

# Add the parent directory to the system path to allow importing of proximitylib
sys.path.append(str(Path("..").resolve()))

import numpy as np
import importlib
import proximitylib.proximity as prox
importlib.reload(prox)
from tabulate import tabulate

# --------------------------------------------------------------
# Function to show results
# --------------------------------------------------------------
def show_results(path, name, params, num_null=1000, alpha=0.05,  models=["expon", "uniform", "resampled"], detailed=False, seed=123):
    """
    Analyzes a dataset peaks table and prints the summarized results.
    Parameters:
        path (str): Path to the dataset peaks table.
        name (str): Name of the dataset for display purposes.
        params (dict): Parameter dictionary for analysis.
        num_null (int): Number of null models to generate.
        alpha (float): Significance level for statistical tests.
        models (list): List of models to consider.
        detailed (bool): Whether to print the detailed peaks table.
    Returns:
        df (DataFrame): Detailed results table.
        df_summary (DataFrame): Summarized results table.
        distance_corr_dict (dict): Dictionary of distance correlations.
    """
    df, distance_corr_dict = prox.analyze_dataset_peaks_table(path, params, num_null=num_null, alpha=alpha, models=models, seed=seed)
    
    if detailed == True:
        prox.print_peak_table(df, title=name)

    df_summary = prox.summarize_results_table(df, distance_corr_dict=distance_corr_dict, alpha=alpha)
    print(f"\n{name} (Summary)\n" + "="*len(f"{name} (Summary)"))
    print(tabulate(df_summary, headers="keys", tablefmt="fancy_grid", showindex=False))

    return df, df_summary, distance_corr_dict

# === Seed for reproducibility ===
SEED = 42 # seed used for the ratio peak analysis in the paper: 123
rng = np.random.default_rng(SEED)

# === Choose a Parameter-Set ===
param_choice = 3         # options 1, 2 or 3 
params = prox.param_sets[param_choice]
print(f"Applied Parameter-Dict: default_params_{param_choice}")

# --------------------------------------------------------------
# Application of generated and empirical datasets
# --------------------------------------------------------------

# Corresponding to Tables 2 and S2
isochrony_small_jitter = show_results("../data/theoretical_sequences/baseIOI-0.50/20beats/isochrony_jitter-0001", "Isochrony Small Jitter", params=params, seed=SEED)
isochrony_big_jitter = show_results("../data/theoretical_sequences/baseIOI-0.50/20beats/isochrony_jitter-0050", "Isochrony Big Jitter", params=params, seed=SEED)
heterochrony_1_2 = show_results("../data/theoretical_sequences/baseIOI-0.50/20beats/heterochrony-1-2_jitter-0050", "Heterochrony 1-2", params=params, seed=SEED)
heterochrony_2_3 = show_results("../data/theoretical_sequences/baseIOI-0.50/20beats/heterochrony-2-3_jitter-0050", "Heterochrony 2-3", params=params, seed=SEED)
tempo_shift = show_results("../data/theoretical_sequences/baseIOI-0.50/20beats/tempo_shift_ioi-jitter-0050", "Tempo Shift", params=params, seed=SEED)
accelerando = show_results("../data/theoretical_sequences/baseIOI-0.50/20beats/accelerando/accelerando.csv", "Accelerando", params=params, seed=SEED)
slow_ritardando = show_results("../data/theoretical_sequences/baseIOI-0.50/20beats/ritardando_slow/ritardando_slow.csv", "Slow Ritardando", params=params, seed=SEED)
jitter_middle = show_results("../data/theoretical_sequences/baseIOI-0.50/20beats/jitter_middle-0050", "Jitter middle", params=params, seed=SEED)
random_exponential = show_results("../data/theoretical_sequences/baseIOI-0.50/20beats/random_exponential", "Random Exponential", params=params, seed=SEED)
random_uniform = show_results("../data/theoretical_sequences/baseIOI-0.50/20beats/random_uniform", "Random Uniform", params=params, seed=SEED)

# Corresponding to Figure 4 and Table S2
kwa = show_results("../data/Kwa-Sounds (Scorpaena spp.)", "Scorpaena spp.", params=params, seed=SEED)
C_persp = show_results("../data/Seba's short-tailed bat (Carollia perspicillata)", "Carollia perspicillata", params=params, seed=SEED)
Sperm_whale = show_results("../data/Sperm whale (Physeter macrocephalus)", "Physeter macrocephalus", params=params, seed=SEED)
```
