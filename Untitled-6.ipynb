```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import gaussian_kde

def plot_ratio_of_normals_with_distributions_cv(
    sample_size=10000,
    mean1=1.0,
    cv1=0.2,
    mean2=1.0,
    cv2=0.2,
    bins=100,
    random_seed=None,
    show_kde=True
):
    """
    Simuliert zwei Normalverteilungen mit gegebenem Mittelwert (mean)
    und Variationskoeffizient (CV = std/mean), zieht je einen Wert aus beiden,
    bildet deren Ratio (größer/kleiner) und zeigt:
      - oben: Histogramme beider Verteilungen
      - unten: Histogramm + KDE der Ratios

    Parameter:
    -----------
    sample_size : int
        Anzahl der gezogenen Werte
    mean1, mean2 : float
        Erwartungswerte der Verteilungen
    cv1, cv2 : float
        Variationskoeffizienten (σ/μ) der Verteilungen
    bins : int
        Anzahl Histogrammbins
    random_seed : int oder None
        Optionaler Zufalls-Seed
    show_kde : bool
        Ob eine KDE-Dichtekurve im Ratio-Plot angezeigt wird
    """

    if random_seed is not None:
        np.random.seed(random_seed)

    # 1️⃣ Standardabweichungen aus CV berechnen
    std1 = mean1 * cv1
    std2 = mean2 * cv2

    # 2️⃣ Zufallswerte aus beiden Normalverteilungen ziehen
    x1 = np.random.normal(mean1, std1, sample_size)
    x2 = np.random.normal(mean2, std2, sample_size)

    # 3️⃣ Ratios bilden (größere / kleinere Zahl)
    ratios = np.maximum(x1, x2) / np.minimum(x1, x2)

    # 4️⃣ Figure mit zwei Subplots
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(8, 8), sharex=False)
    fig.suptitle(
        f"Ratio-Verteilung zweier Normalverteilungen\n"
        f"N₁(μ={mean1}, CV={cv1}), N₂(μ={mean2}, CV={cv2})",
        fontsize=13
    )

    # -------------------------------
    # OBERER PLOT: Zwei Normalverteilungen
    # -------------------------------
    ax1.hist(x1, bins=bins, alpha=0.5, density=True, color="steelblue", label=f"N₁(μ={mean1}, CV={cv1})")
    ax1.hist(x2, bins=bins, alpha=0.5, density=True, color="darkorange", label=f"N₂(μ={mean2}, CV={cv2})")
    ax1.set_ylabel("Dichte", fontsize=11)
    ax1.set_title("Verteilungen der gezogenen Werte", fontsize=12)
    ax1.legend()
    ax1.grid(alpha=0.3)

    # -------------------------------
    # UNTERER PLOT: Ratios + KDE
    # -------------------------------
    ax2.hist(ratios, bins=bins, density=True, alpha=0.4, color="slateblue", label="Ratios (Histogramm)")
    if show_kde:
        kde = gaussian_kde(ratios)
        x_grid = np.linspace(1, np.percentile(ratios, 99.9), 1000)
        ax2.plot(x_grid, kde(x_grid), color="darkblue", lw=2, label="KDE-Dichte")
    ax2.set_xlabel("Ratio (größer/kleiner)", fontsize=11)
    ax2.set_ylabel("Dichte", fontsize=11)
    ax2.set_title("Verteilung der Ratios", fontsize=12)
    ax2.legend()
    ax2.set_xlim(1, np.percentile(ratios, 99.9))
    ax2.grid(alpha=0.3)

    plt.tight_layout(rect=[0, 0, 1, 0.96])
    plt.show()

    return x1, x2, ratios

plot_ratio_of_normals_with_distributions_cv(
    sample_size=10000,
    mean1=1.2, 
    cv1=0.1, 
    mean2=1.0, 
    cv2=0.3, 
    random_seed=42)
```
